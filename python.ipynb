```python
#identity operators
a = None

print(a is None)
print(a is not None)
```

```python
#bitwise operators
a = 5
b = 3

print(a & b)
print(a | b)
print(a ^ b)
```

```python
#electricity bill calculator
unit = int(input("Enter the number of units consumed: "))
rate =6
bill = unit * rate
print("The electricity bill is: ", bill)
```

```python
#travel expense calculator
travel_expense = float(input("Enter the total travel expense: "))
food_expense = float(input("Enter the total food expense: "))
hotel_expense = float(input("Enter the total hotel expense: "))
total_expense = travel_expense + food_expense + hotel_expense   
print("The total travel expense is: ", total_expense)
```
