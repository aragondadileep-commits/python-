```python
#list on python
#list is an ordered and changeable collection that can store the different data types

marks =[80, 90, 75, 85]

print(marks)[0]
print(marks[1])
print(marks[2])
print(marks[3])
```

```python
#change elements in the list
marks[1]=95
print(marks)
```

```python
#add elements in the list
marks=[80, 90,75]
marks.append(85)
print(marks)
```

```python
#remove elements from a list
marks =[80, 90, 75]

marks.remove(90)
print(marks)
```
